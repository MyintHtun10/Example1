```python

import pandas as pd

data = {
    "Item": ["Apple", "Banana", "Coconut", "Durian", "Papaya"],
    "Quantity": [7, 2, 5, 1, 8],
    "Price": [7500, 5000, 3000, 80000, 4000]
}

df = pd.DataFrame(data)

df["Amount"] = df["Quantity"] * df["Price"]

df["Remark"] = df["Amount"].apply(
    lambda x: "Discount" if x > 50000 else "No Discount"
)

df
expensive = df.loc[df["Amount"].idxmax()]


print("Top Exensive:",expensive["Item"])
print("Charges:", expensive["Amount"])

cheap=df.loc[df['Amount'].idxmin()]
print("Top Cheap:",cheap["Item"])
print("Charges:",cheap["Amount"])

```
